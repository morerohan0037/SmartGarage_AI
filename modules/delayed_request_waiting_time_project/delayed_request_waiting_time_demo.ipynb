# Delayed Request Detection & Waiting-Time Prediction

Seeded synthetic ML prototype for garage-door and automatic-gate field service.

Tasks:
1. predict waiting time in minutes,
2. predict whether a request will exceed its urgency SLA.


## Run the complete pipeline
The full executable script is `run_delayed_request_project.py`. From the project folder run:

```bash
python -m pip install -r requirements.txt
python run_delayed_request_project.py
```

For a single example request, run `python predict_new_request.py` after training.
